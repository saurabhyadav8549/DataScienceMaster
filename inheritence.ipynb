```python
class test:        
    def test_meth(self):
        return "this is my first class"
```

```python
class child_test(test):
    pass
```

```python
child_test_obj = child_test()
```

```python
child_test_obj.test_meth()
```

```python
class class1:  ## multi level inheritence

    def test_classs1(self):
        return "this is method for class1"
```

```python
class class2(class1) :
    def test_class2(self):
        return "this is method for class 2"
    
```

```python
class class3(class2):
    pass
```

```python
obj_class3= class3()
```

```python
obj_class3.test_classs1()
```

```python
obj_class3.test_class2()
```

```python
# multiple inheritence
```

```python
class class1:
    def test_class1(self):
        return "this is class 1"
```

```python
class class2:
    def test_class2(self):
        return "this is class 2"
```

```python
class class3(class1,class2):
    pass
```

```python
obj_class3= class3()
```

```python
obj_class3.test_class1()
```

```python
obj_class3.test_class2()
```
